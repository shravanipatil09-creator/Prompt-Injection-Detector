```python

import json
from datetime import datetime
```

```python
def create_log(text, result, event_id):

    log = {
        "event_id": event_id,
        "timestamp": str(datetime.now()),
        "input": text,
        "decision": result["decision"],
        "score": result["score"],
        "severity": result["severity"],
        "reasons": result["reasons"]
    }

    return log
```

```python

log = {
    "event_id": 1,
    "timestamp": str(datetime.now()),
    "input": "ignore previous instructions",
    "decision": "Block",
    "score": 2,
    "severity": "Medium",
    "reasons": [
        "ignore previous instructions"
    ]
}

print(log)
```

```python
logs = [
    {
        "event_id": 1,
        "timestamp": str(datetime.now()),
        "input": "ignore previous instructions",
        "decision": "Block",
        "score": 2,
        "severity": "Medium",
        "reasons": ["ignore previous instructions"]
    },
    {
        "event_id": 2,
        "timestamp": str(datetime.now()),
        "input": "api_key=abcd1234",
        "decision": "Block",
        "score": 2,
        "severity": "High",
        "reasons": ["api_key"]
    },
    {
        "event_id": 3,
        "timestamp": str(datetime.now()),
        "input": "Hello, how are you",
        "decision": "Allow",
        "score": 0,
        "severity": "Low",
        "reasons": ["Normal input"]
    }
]

print(logs)
```

```python
with open("logs.json", "w") as file:
    json.dump(logs, file, indent=4)

print("logs.json saved successfully")
```
