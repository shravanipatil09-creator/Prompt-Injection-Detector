```python

# 2) Imports and helper functions
import os
import re
import regex
import unicodedata
from unidecode import unidecode
import pandas as pd

def normalize_text(s):
    s = "" if s is None else str(s)
    s = unicodedata.normalize("NFKC", s)
    # remove zero-width and control characters
    s = regex.sub(r'[\p{Cf}\p{Cc}]', '', s)
    return s

def to_ascii(s):
    return unidecode(normalize_text(s))
```

```python
from unidecode import unidecode

def to_ascii(text):
    if not isinstance(text, str):
        text = str(text)
    return unidecode(text)
```

```python
df["normalized"] = df["input"].astype(str).apply(normalize_text)
df["ascii"] = df["normalized"].apply(to_ascii)

out_path = "/content/review_batch.csv"
df.to_csv(out_path, index=False)

print("Saved:", out_path)
df.head()
```
